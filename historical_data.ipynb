```python

import pandas as pd

sales = {
    2022: [
        826.80, 760.20, 831.21, 803.56,
        847.54, 849.10, 862.05, 826.82,
        855.34, 862.00, 854.70, 829.06
    ],
    2023: [
        870.84, 828.70, 847.47, 844.34,
        873.24, 850.53, 881.01, 885.17,
        874.35, 877.48, 859.43, 828.34
    ]
}

rows = []

for year, values in sales.items():
    for month_num, value in enumerate(values, start=1):
        rows.append({
            "date": f"{year}-{month_num:02d}-01",
            "year": year,
            "month_num": month_num,
            "kplc_sales_gwh": value,
            "source": (
                "KNBS Leading Economic Indicators, "
                "October 2024, Table 15(b)"
            ),
            "verification_status": "Candidate - reconcile report vintages"
        })

historical = pd.DataFrame(rows)
historical["date"] = pd.to_datetime(historical["date"])

historical.to_csv(
    "powerpulse_historical_2022_2023_candidates.csv",
    index=False
)

print("Historical candidate file created.")
print("Rows:", len(historical))
print("Date range:", historical["date"].min(),
      "to", historical["date"].max())
print(historical.head())
print(historical.tail())
```
